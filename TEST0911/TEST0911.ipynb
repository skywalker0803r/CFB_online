## 擷取數據

```python
from datetime import datetime, timedelta
# 計算近 50 天的時間範圍
end_time = datetime.now()  # 當前時間
start_time = end_time - timedelta(days=50)  # 50 天前

# 將時間轉換為適當的格式（取決於 async_get_values 的 API 要求）
start_time_str = start_time.strftime('%Y-%m-%d %H:%M:%S')
end_time_str = end_time.strftime('%Y-%m-%d %H:%M:%S')
```

```python
#help(async_get_values)
```

```python
# 獲取數據 CFB1
cfb1 = async_get_values(
    tag1.keys(),
    timestart=start_time_str,
    timeend=end_time_str,
    interval='10s'  # 10 秒間隔
)

# 後續處理（與原程式碼相同）
for c in cfb1:
    cfb1[c] = pd.to_numeric(cfb1[c], errors='coerce')

cfb1['前爐SOx濃度'] = cfb1['MLUT4_FIQ-1BTCF'] * 8 / cfb1['MLUT4_FQ-139'] * 24.5 * 1000 / 32.065
cfb1['鈣硫比'] = cfb1['MLUT4_RQ-1BTLS'] / cfb1['MLUT4_FIQ-1BTCF'] * 0.32 / 0.008
cfb1['DeSOx_1st'] = (cfb1['前爐SOx濃度'] - cfb1['MLUT4_AIC-132B']) / cfb1['前爐SOx濃度']
cfb1['DeSOx_2nd'] = (cfb1['MLUT4_AIC-132B'] - cfb1['MLUT4_AT-140']) / cfb1['MLUT4_AIC-132B']

cfb1 = cfb1[sorted(cfb1.columns)]
new_index = cfb1.index.to_list() + ['datetime']
cfb1 = pd.concat([pd.Series(tag1).to_frame().T, cfb1], axis=0, ignore_index=True)
cfb1.index = new_index
```
